# Sampling from DataFrames

### Random Sampling
`data.sample()` → randomly selects rows from a DataFrame.

`data.sample(n=5)` → randomly selects 5 rows.

`replace=False` → a row cannot be selected more than once.

`replace=True` → the same row can be selected multiple times.

### Selecting Data with `.iloc`

Syntax: `data.iloc[rows, columns]`

`:` → selects all rows.

`-3:` → selects the last 3 columns.

Example:

`sample.iloc[:, -3:]`

→ selects **all rows and the last 3 columns**.

### Complete Example

`sample = data.sample(n=5, replace=False)`

`print(sample.iloc[:, -3:])`

→ Randomly selects 5 different rows and displays their last 3 columns.

### SCATTER PLOT WITH MULTIPLE LAYERS

````python
plt.plot(data.sepalLength, data.sepalWidth, ls='', marker='o', label='sepal')
plt.plot(data.petalLength, data.petalWidth, ls='', marker='o', label='petal')
plt.legend()
````

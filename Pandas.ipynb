# Analysing data

### viewing the data

One of the most used method for getting a quick overview of the DataFrame, is the head() method.

The head() method returns the headers and a specified number of rows, starting from the top.

```python
import pandas as pd
df =pd.read_json('iris.json')
df.head(10)
#prints first 10 elements of dataframe
```

```python
import pandas as pd
df =pd.read_json('iris.json')
df.tail(10)
#prints last 10 elements of dataframe
```

```python
import pandas as pd
df =pd.read_json('iris.json')
df.head()
#prints first 5 elements of dataframe if only head()
```
